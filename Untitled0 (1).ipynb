```python
from textblob import TextBlob

sample_review = df['review'][0]
blob = TextBlob(sample_review)
print(blob.sentiment)
```

```python
def get_sentiment(text):
    blob = TextBlob(text)
    if blob.sentiment.polarity > 0:
        return 'positive'
    else:
        return 'negative'

df_sample = df.head(100).copy()
df_sample['predicted_sentiment'] = df_sample['review'].apply(get_sentiment)
df_sample[['review', 'sentiment', 'predicted_sentiment']].head(10)
```

```python
accuracy = (df_sample['sentiment'] == df_sample['predicted_sentiment']).mean() * 100
print(f"Accuracy: {accuracy}%")
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://quotes.toscrape.com/"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

print(soup.title.text)
```

```python
quotes = soup.find_all('span', class_='text')
authors = soup.find_all('small', class_='author')

for quote, author in zip(quotes, authors):
    print(quote.text, "-", author.text)
```
